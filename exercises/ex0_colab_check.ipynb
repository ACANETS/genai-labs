## 2. GPU vs. CPU: a matrix multiplication
Deep learning is mostly large matrix multiplications. Compare the time for a 4096 x 4096 product on the CPU and on the GPU.

```python
import torch
a = torch.randn(4096, 4096)
b = torch.randn(4096, 4096)
with measure("CPU matmul (float32)"):
    c = a @ b
if device != "cpu":
    a_d, b_d = a.to(device, dtype), b.to(device, dtype)
    _ = a_d @ b_d                       # warm-up
    with measure(f"{device} matmul ({str(dtype).replace('torch.', '')})"):
        for _ in range(10):
            c_d = a_d @ b_d
        if device == "cuda":
            torch.cuda.synchronize()
    print("(the GPU time above is for 10 products)")
```
