```python
import importlib.util
if importlib.util.find_spec("genai_utils") is None:
    get_ipython().run_line_magic("pip", 'install -q "git+https://github.com/acanets/genai-labs@spring27"')
from transformers import pipeline
from genai_utils import setup_env
device, _ = setup_env(need_gpu=False)
clf = pipeline("text-classification", model="facebook/roberta-hate-speech-dynabench-r4-target",
               device=0 if device == "cuda" else -1, top_k=None)
```

## 1. Score a few texts

```python
texts = [
    "I really enjoyed this week's lecture on diffusion models.",
    "This homework is so hard it's killing me.",
    "People from that town are all lazy and dishonest.",
    "Thanks for answering my question during the chat!",
]
for t in texts:
    scores = {d["label"]: round(d["score"], 3) for d in clf(t)[0]}
    print(f"{scores}  <- {t}")
```
